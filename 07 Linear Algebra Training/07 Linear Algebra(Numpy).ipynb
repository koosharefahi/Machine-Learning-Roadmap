# Linear Algebra with NumPy: Solving a Multivariate Linear System

## Introduction

The goal of this project is to explore fundamental linear algebra concepts and perform mathematical operations using Python's NumPy library. A synthetic dataset will be created to practice working with vectors and matrices, without relying on external datasets.

Throughout this project, we will cover the following operations:

- **Vector Creation:** Creating and working with one-dimensional arrays.
- **Matrix Creation:** Defining matrices and inspecting their dimensions.
- **Vector Addition:** Performing element-wise addition between vectors.
- **Dot Product:** Calculating the inner product of two vectors.
- **Matrix Multiplication:** Multiplying matrices and exploring the importance of matrix dimensions and multiplication order.
- **Vector Norms:** Measuring the magnitude of vectors.
- **Matrix Transpose:** Interchanging the rows and columns of a matrix.
- **Linear System Construction:** Representing a multivariate system of linear equations in matrix form (\(Ax=b\)).
- **Solving Linear Systems:** Using `numpy.linalg.solve()` to calculate the unknown variables.
- **Solution Verification:** Checking the calculated solution by substituting it back into the original system.

By the end of this project, we will have a practical understanding of basic linear algebra operations and how NumPy can be used to solve systems of linear equations efficiently.

```python
import numpy as np
```

```python
# Create Synthetic Dataset
# Create Vectors
a = np.array([2,4,6])
b = np.array([1,3,5])
```

```python
print(a)
```

```python
# Create Matrices
m_a = np.array([[1,2,3] ,
              [4,5,6]])

m_b = np.array([[7,8,9],
               [10,11,12]])
```

```python
print(m_a)
print(m_a.shape)
print(m_a.ndim)
```

```python
print(m_b)
print(m_b.shape)
print(m_b.ndim)
```

```python
# Implement Vector Addition
v_sum = a + b
v_sum
```

```python
m_sum = m_a + m_b
m_sum
```

```python
# Dot products
v_dot = np.dot(a,b)
print(v_dot)
```

```python
# both matrices are (2,3), can not implement multiplication
m_b_t = np.transpose(m_b)
m_b_t
```

```python
m_a_prod = m_a @ m_b_t
print(m_a_prod)
print("")
m_b_prod = m_b_t @ m_a 
print(m_b_prod)
```

```python
# Vector Norms
# v = radical(v1^2 + v2^2)

norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

print(norm_a)
print(norm_b)
```

```python
# Transpose Matrice A
m_a_t = np.transpose(m_a)

print(m_a)
print("")
print(m_a_t)
print("")
print(m_a.shape)
print(m_a_t.shape)
```

```python
# Create a Linear System
c = np.array([[2,1,1] ,
             [1,3,2],
             [3,2,2]])
d = np.array([[8]
              ,[13]
              ,[16]])

print(c)
print("")
print(d)
print("")
print(c.shape)
print(d.shape)
```

```python
# Solve the system cx = d
solve = np.linalg.solve(c,d)
```

```python
print(np.round(solve, 6))
```

```python
verification = c @ solve
```

```python
print(np.allclose(verification, d))
```
