### Calculate mean, median, variance, std & Compare with built-in functions
$$\mu = \frac{1}{N}\sum_{i=1}^{N} x_i \quad , \quad \sigma^2 = \frac{1}{N}\sum_{i=1}^{N} (x_i - \mu)^2 \quad , \quad \sigma = \sqrt{\sigma^2}$$

```python
N = len(sales)
```

```python
# Calculate Mean
mean = sum(sales)/N
mean2 = np.mean(sales)
print(mean)
print(mean2)
```

```python
# Calculate Median
sorted_sales = np.sort(sales)
if N % 2 == 1 :
    median = sorted_sales[N//2]
else:
    median = (sorted_sales[N//2 - 1] + sorted_sales[N//2]) / 2

numpy_median = np.median(sales)
```

```python
print(median)
print(numpy_median)
```

```python
# Calculate Variance and STD
variance = np.sum((sales - mean)**2)/N
numpy_var = np.var(sales)
print(variance)
print(numpy_var)
```

```python
# Calculate STD
std = np.sqrt(variance)
numpy_std = np.std(sales)
print(std)
print(numpy_std)
```

```python
print(f"Mean   -> Custom: {mean:.4f} | NumPy: {mean2:.4f} | Match: {np.isclose(mean, mean2)}")
print(f"Median -> Custom: {median:.4f} | NumPy: {numpy_median:.4f} | Match: {np.isclose(median, numpy_median)}")
print(f"Var    -> Custom: {variance:.4f} | NumPy: {numpy_var:.4f} | Match: {np.isclose(variance, numpy_var)}")
print(f"Std    -> Custom: {std:.4f} | NumPy: {numpy_std:.4f} | Match: {np.isclose(std, numpy_std)}")
```

### (Z-Score Normalization) و Min-Max Scaling
$$Z = \frac{X - \mu}{\sigma} \quad , \quad X_{\text{norm}} = \frac{X - X_{\min}}{X_{\max} - X_{\min}}$$

```python
# Implement Z-Normalization
z_score_norm = (sales - mean)/std
print(z_score_norm)
print(f"Z-Normalized Mean: {np.mean(z_score_norm):.2f} (Expected: 0.00)")
print(f"Z-Normalized Std : {np.std(z_score_norm):.2f} (Expected: 1.00)")
```

```python
# Scaling
max_val = np.max(sales)
min_val = np.min(sales)
print(max_val)
print(min_val)
minmax_scale = (sales - min_val) / (max_val - min_val)
print(minmax_scale)
```

### Build a function for statistical summaries

```python
def statistics_sum(arr , name='Array') :
    arr = np.asanyarray(arr , dtype=np.float64)
    n = len(arr)

    #vectorized
    mean_cal = np.sum(arr)/n
    var_cal = np.sum((arr - mean_cal)**2)/n
    std_cal = np.sqrt(var_cal)

    q25, median_cal, q75 = np.percentile(arr, [25,50,75])
    iqr = q75 - q25


    summary={
        "Count" : n,
        "Mean" : mean_cal,
        "Variance" : var_cal,
        'STD' : std_cal ,
        'Min' : np.mean(arr),
        "25% (Q1)" : q25,
        "50% (Median)" : median_cal,
        "75% (Q3)" : q75,
        'IQR' : iqr,
        'Max' : np.max(arr),
        "Sum" : np.sum(arr)
    }

    print(f"Statistical Summary Engine: {name}")
    print(" " * 45)
    for k, v in summary.items():
        if isinstance(v, float):
            print(f"{k:<15}: {v:,.4f}")
        else:
            print(f"{k:<15}: {v:,}")
    print(" " * 45)
    return summary
```

```python
stats_report = statistics_sum(sales, name="Superstore Sales")
```
